```python
import pandas as pd
import numpy as np
import matplotlib

data = 'data/employee_count_by_dep.csv'
df = pd.read_csv(data)

display(df)
print(df.columns)
```

```python
departments = df['Department Name']
em_count = df['Employee Count']

plt.barh(departments, em_count, color='green')

for index, value in enumerate(em_count):
    plt.text(value + 1, index - 0.2, str(value))

plt.xlabel('Employees count')
plt.ylabel('Departments Names')
plt.title("Employee count in each department")

plt.show()
```

```python
plt.plot(em_count, departments, marker='o')
plt.xlabel('Departments')
plt.ylabel('Employee count')
plt.title('Count of employee in each department')
plt.xlim(1)
plt.show()
```
