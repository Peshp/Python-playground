```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([2020, 2021, 2022, 2023, 2024, 2025, 2026])
y = np.array([29.8, 34.62, 64.5, 67.1, 38, 99.8, 56.4])

plt.plot(x, y, color='red', marker='o')

y2 = np.array([19.8 ,31, 66.1, 78.7, 83, 96, 55.1])
plt.plot(x, y2, color='green', marker='o')

plt.title('Values per year', weight='bold', family='Calibri')
plt.xlabel('Year', family='Arial')
plt.ylabel('Value', family='Times New Roman')

y3 = np.random.randint(0, 600, size=7)
plt.plot(x, y3, color='blue', marker='o')

plt.grid(axis='x')
plt.grid(axis='y', color='lightgrey', linewidth='2')

plt.show()
```
