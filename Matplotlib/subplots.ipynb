```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(1, 10, 11)
y = x ** 4

a = np.arange(0, 11)
b = a * x
```

```python
fig, axes = plt.subplots()

axes.plot(x, y)
axes.set_title('Random Plot')
axes.set_xlabel("x label")
axes.set_ylabel('y label')
```

```python
fig, axes = plt.subplots(1, 2)

for a in axes:
    a.plot(y, b)
```

```python
fig, axes = plt.subplots(2, 2)

axes[0, 1].plot(x, y)
axes[1, 0].plot(b, b)

plt.tight_layout()
```

```python
fig, axes = plt.subplots(2, 2, figsize=(5, 3))

axes[0, 0].plot(x, y)
axes[0, 1].plot(y, x)
axes[1, 0].plot(y, b)
axes[1, 1].plot(b, y)

fig.subplots_adjust(left=None,
                    bottom=None,
                    right=None,
                    top=None,
                    wspace=0.9,
                    hspace=0.1)

fig.savefig('random_plots', bbox_inches='tight')
```
